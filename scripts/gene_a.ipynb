**QUESTION 2A**

```python
with open("../data/genea.fasta", "r") as data: #open fasta file for reading 
    cds_sequence = ""     #empty string to store mRNA seqeunce
    for line in data:
        if line[0] != ">":   #skip the fasta header line
            cds_sequence += line.replace("\n", "")  #to remove the new lines and add the sequence to the string
        
cds = cds_sequence[147:1494]  #extract the CDS sequence (index starts from 0, so 148 is index 147)

print("Length of the CDS sequence:", len(cds))

# Check divisibility by 3
if len(cds) % 3 == 0:
    print("CDS length is divisible by 3")
    print("Number of amino acids:", round(len(cds)/3))
else:
    print("CDS length is NOT divisible by 3; Check the CDS coordinates or sequence.")
```

**QUESTION 2B**

```python
# dictionary containing genetic code
codon_table = {
"ATT":"I", "ATC":"I", "ATA":"I", "ATG":"M","ACT":"T", "ACC":"T", "ACA":"T", "ACG":"T",
"AAT":"N", "AAC":"N", "AAA":"K", "AAG":"K","AGT":"S", "AGC":"S", "AGA":"R", "AGG":"R",

"GTT":"V", "GTC":"V", "GTA":"V", "GTG":"V","GCT":"A", "GCC":"A", "GCA":"A", "GCG":"A",
"GAT":"D", "GAC":"D", "GAA":"E", "GAG":"E","GGT":"G", "GGC":"G", "GGA":"G", "GGG":"G",

"TTT":"F" , "TTC":"F", "TTA":"L", "TTG":"L", "TCT":"S", "TCC":"S", "TCA":"S", "TCG":"S",
"TAT":"Y", "TAC":"Y", "TAA":"*", "TAG":"*", "TGT":"C", "TGC":"C", "TGA":"*", "TGG":"W",

"CTT":"L", "CTC":"L", "CTA":"L", "CTG":"L", "CCT":"P", "CCC":"P", "CCA":"P", "CCG":"P",
"CAT":"H", "CAC":"H", "CAA":"Q", "CAG":"Q","CGT":"R", "CGC":"R", "CGA":"R", "CGG":"R",
}
sephs2_protein=""   # protein string is defined
for i in range(0, len(cds), 3): # Read CDS three nucleotides (one codon) at a time
    sephs2_protein+=codon_table[cds[i:i+3]] # convert each codon into its mapped amino acid in dictionary

print("First 30 residues of the protein:", sephs2_protein[0:30]) # print the first 30 residues
print(f"Total length of the protein:{len(sephs2_protein)} aa") # print the total length (aa)
```

**QUESTION 2C**

```python
with open("../data/genea_pro.fasta") as f:      # Open the deposited protein FASTA file
    selenide_protein = ""              # string to store the protein sequence
    for line in f: 
        if line[0] != ">":             # To exclude fasta header
            selenide_protein += line.replace("\n", "") # Add sequence lines without line breaks

    if sephs2_protein.endswith("*"): #remove the stop codon before comparing
        sephs2_protein_clean = sephs2_protein[:-1]
    else:
        sephs2_protein_clean = sephs2_protein
        
# Compare the sequences
if sephs2_protein_clean == selenide_protein:    # check both ssequences are identical 
    print("Sequences are identical.")
else:
    if len(sephs2_protein_clean) == len(selenide_protein):
            for i in range(len(sephs2_protein_clean)):  # compare against each amino acid position in loop 
                if sephs2_protein_clean[i] != selenide_protein[i]: #  check amino acid positon with thier index
                    print("Difference position:", i + 1)        # first difference in thier position; +1 is added since index starts at 0
                    print("Translated protein amino acid:", sephs2_protein_clean[i])        # amino acid od the translated protein
                    print("Selenide (deposited) protein amino acid:", selenide_protein[i])    # amino acid of the deposited protein
```

**QUESTION 2D**

```python
from Bio.Seq import Seq   # Import Seq class from biopython
sequence = Seq(cds)       # convert CDS DNA into biopython Seq object
bio_trans_protein= sequence.translate()   # translate the CDS into protein
if sephs2_protein == bio_trans_protein:   # compare the manually translated and biopython translated protein
    print("Both the sequences are identical")
else: 
    print("There is a difference in translation")

print("Biopython translated  protein: ", bio_trans_protein)  # biopython translated protein
print("Manually translated protein: ", sephs2_protein)  # manually translated protein
```
