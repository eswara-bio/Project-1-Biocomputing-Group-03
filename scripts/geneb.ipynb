```python
with open("../data/geneb.fasta") as fd:
    lines = fd.read().splitlines()
    seq = ''.join([line for line in lines if not line.startswith('>')])

print(f"Length of whole sequence: {len(seq)}")
```

```python
# 1-based indexing
cds_start = 10059
cds_end   = 10404

# 0-based indexing and including the last codon
gene = seq[cds_start - 1 : cds_end]

print(f"Length of gene mRNA: {len(gene)}")
print(f"Is the length of gene divisible by 3? : {not (len(gene) % 3)}")
```

```python
gene[-6:]
```




The note in the CDS part of gene specifically mentions - 

""
TAA stop codon is completed by the addition of 3' A residues to the mRNA
""

So we can discard the last nucleotide which is part of stop codon.

```python
gene = gene[:-1]
print("Gene:", gene)

print(f"Length of gene: {len(gene)}")
print(f"Is the length of gene divisible by 3? : {not (len(gene) % 3)}")
```

# Q2 b)

```python
# skipping stop codons
reverse_codes = {
    'F': ['TTT', 'TTC'],
    'L': ['TTA', 'TTG', 'CTT', 'CTC', 'CTA', 'CTG'],
    'I': ['ATT', 'ATC'],
    'M': ['ATA', 'ATG'],
    'V': ['GTT', 'GTC', 'GTA', 'GTG'],
    'S': ['TCT', 'TCC', 'TCA', 'TCG', 'AGT', 'AGC'],
    'P': ['CCT', 'CCC', 'CCA', 'CCG'],
    'T': ['ACT', 'ACC', 'ACA', 'ACG'],
    'A': ['GCT', 'GCC', 'GCA', 'GCG'],
    'Y': ['TAT', 'TAC'],
    'H': ['CAT', 'CAC'],
    'Q': ['CAA', 'CAG'],
    'N': ['AAT', 'AAC'],
    'K': ['AAA', 'AAG'],
    'D': ['GAT', 'GAC'],
    'E': ['GAA', 'GAG'],
    'C': ['TGT', 'TGC'],
    'W': ['TGA', 'TGG'],
    'R': ['CGT', 'CGC', 'CGA', 'CGG'],
    'G': ['GGT', 'GGC', 'GGA', 'GGG'],
#    'STOP': ['TAA', 'TAG', 'AGA', 'AGG']
}


codes = {}

for key in reverse_codes:
    for codon in reverse_codes[key]:
        codes[codon] = key

print(len(codes))
```

```python
# now tranlate the sequence to protein
protein_seq = ""
for i in range(0, len(gene), 3):
        protein_seq += codes[gene[i:i+3]]

print(protein_seq)
```

```python
print(f"First 30 residues of translation: {protein_seq[:30]}")
```

```python
with open("../data/geneb_pro.fasta") as fd:
    lines = fd.read().splitlines()
    orig_protein_seq = ''.join([line for line in lines if not line.startswith('>')])

assert protein_seq == orig_protein_seq, "Proteins dont match"
```

No difference between the translated and deposited proteins were found

```python
from Bio.Seq import Seq

dna_seq = Seq(gene)
translated_seq = dna_seq.translate(table=2)

print(translated_seq)

assert translated_seq == protein_seq, "Biopython and our translation does not match"
```

Since the assert condition is true, the biopython translated protein sequence and our translated sequence are identical.
